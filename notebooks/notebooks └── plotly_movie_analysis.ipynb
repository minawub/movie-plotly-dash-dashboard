```python
# =====================================================
# 📚 IMPORT REQUIRED LIBRARIES
# =====================================================

import pandas as pd
import numpy as np
import plotly.express as px
import warnings

warnings.filterwarnings("ignore")
print("✅ Libraries imported successfully")
print("="*70)
# =====================================================
# 📂 LOAD MOVIE DATASET
# =====================================================
movies = pd.read_csv("../data/movies.csv")
print("✅ Dataset loaded successfully")
print(movies.head())
print("="*70)
# =====================================================
# 🔍 DISPLAY FIRST 5 ROWS
# =====================================================
print("First 5 Movies:")
print(movies.head())
print("="*70)
# =====================================================
# 📏 DATASET SIZE
# =====================================================

print("Number of Rows:", movies.shape[0])
print("Number of Columns:", movies.shape[1])
print("="*70)
# =====================================================
# 🏷️ DISPLAY COLUMN NAMES
# =====================================================
print("Dataset Columns:")
print(movies.columns.tolist())
print("="*70)
# =====================================================
# 🔤 DISPLAY DATA TYPES
# =====================================================
print("Column Data Types:")
print(movies.dtypes)
print("="*70)
# =====================================================
# 🧹 CHECK MISSING VALUES
# =====================================================
print("Missing Values:")
print(movies.isnull().sum())
print("="*70)
# =====================================================
# 🧹 REMOVE MISSING VALUES
# =====================================================
movies = movies.dropna()
print("✅ Missing values removed")
print("Remaining Rows:", movies.shape[0])
print("="*70)
```

```python
# =====================================================
# 🎭 EXERCISE 2: MOVIE COUNT BY GENRE
# =====================================================
# Count movies for each genre
genre_count = (movies['genre'].value_counts().reset_index())
genre_count.columns = ["Genre","Movie Count"]
print("🎭 Movies by Genre:")
print(genre_count)
print("="*70)
# =====================================================
# 📊 CREATE GENRE BAR CHART
# =====================================================
fig = px.bar(genre_count,x="Genre",y="Movie Count",color="Genre",
    title="🎭 Number of Movies by Genre"
)
fig.update_layout(
    xaxis_title="Movie Genre",
    yaxis_title="Number of Movies",
    template="plotly_white"
)
fig.show()
```

```python
# =====================================================
# ⭐ EXERCISE 3: AVERAGE SCORE BY GENRE
# =====================================================
# Calculate average score for each genre
genre_score = (movies.groupby("genre")["score"].mean().reset_index())
genre_score.columns = [
    "Genre",
    "Average Score"
]
print("⭐ Average Score by Genre:")
print(genre_score)
print("="*70)
# =====================================================
# 📊 CREATE SCORE BAR CHART
# =====================================================
fig = px.bar(
    genre_score,
    x="Genre",
    y="Average Score",
    color="Average Score",
    title="⭐ Average Movie Score by Genre",
    color_continuous_scale="Viridis"
)
fig.update_layout(
    xaxis_title="Movie Genre",
    yaxis_title="Average Score",
    template="plotly_white"
)
fig.show()
```

```python
# =====================================================
# 💰 EXERCISE 4: BUDGET VS GROSS REVENUE
# =====================================================
fig = px.scatter(
    movies,
    x="budget",
    y="gross",
    color="genre",
    hover_name="name",
    title="💰 Movie Budget vs Gross Revenue"
)
fig.update_layout(
    xaxis_title="Production Budget ($)",
    yaxis_title="Gross Revenue ($)",
    template="plotly_white"
)

fig.show()
```

```python
# =====================================================
# 📈 EXERCISE 5: MOVIES RELEASED PER YEAR
# =====================================================
# Count movies released each year
movies_year = (
    movies['year']
    .value_counts()
    .sort_index()
    .reset_index()
)
movies_year.columns = ["Year","Movie Count"]
print("📅 Movies Released Per Year:")
print(movies_year.head())
print("="*70)
# =====================================================
# 📊 CREATE LINE CHART
# =====================================================
fig = px.line(
    movies_year,
    x="Year",
    y="Movie Count",
    markers=True,
    title="📈 Number of Movies Released Each Year"
)
fig.update_layout(
    xaxis_title="Release Year",
    yaxis_title="Number of Movies",
    template="plotly_white"
)
fig.show()
```

```python
# =====================================================
# 📊 EXERCISE 6: MOVIE SCORE DISTRIBUTION
# =====================================================
fig = px.histogram(
    movies,
    x="score",
    nbins=20,
    title="📊 Distribution of Movie Scores"
)
fig.update_layout(
    xaxis_title="Movie Score",
    yaxis_title="Number of Movies",
    template="plotly_white"
)
fig.show()
```

```python
# =====================================================
# 🥧 EXERCISE 7: MOVIE RATING DISTRIBUTION
# =====================================================
rating_count = (
    movies['rating']
    .value_counts()
    .reset_index()
)
rating_count.columns = [
    "Rating",
    "Movie Count"
]
print("🎬 Movie Ratings:")
print(rating_count)
print("="*70)
# =====================================================
# 🥧 CREATE PIE CHART
# =====================================================
fig = px.pie(
    rating_count,
    names="Rating",
    values="Movie Count",
    title="🥧 Movie Rating Distribution"
)
fig.update_layout(
    template="plotly_white"
)
fig.show()
```

```python
# =====================================================
# 🏆 EXERCISE 8: TOP 10 HIGHEST GROSSING MOVIES
# =====================================================
top_gross_movies = (movies.sort_values(by="gross",ascending=False).head(10))
print("🏆 Top 10 Highest Grossing Movies:")
print(top_gross_movies[["name","gross"]])
print("="*70)
# =====================================================
# 📊 HORIZONTAL BAR CHART
# =====================================================
fig = px.bar(
    top_gross_movies,
    x="gross",
    y="name",
    orientation="h",
    color="gross",
    title="🏆 Top 10 Highest Grossing Movies"
)
fig.update_layout(
    xaxis_title="Gross Revenue ($)",
    yaxis_title="Movie Name",
    template="plotly_white"
)

fig.show()
```
